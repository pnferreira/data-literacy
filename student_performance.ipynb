## Quartiles, Percentiles, IQR and Boxplot

Before looking for unusual values, we need to understand how the data is distributed.

A **percentile** indicates the value below which a certain percentage of the data falls.

For example:

- The 25th percentile means that 25% of the values are below that point.
- The 50th percentile means that 50% of the values are below that point.
- The 75th percentile means that 75% of the values are below that point.
- The 99th percentile means that 99% of the values are below that point.

In class environments, this is talked about a lot: "the student was in the 90th percentile", meaning it was better than 90% of the class.

Quartiles are specific percentiles that divide the data into four parts:

- `Q1`: first quartile, equivalent to the 25th percentile
- `Q2`: second quartile, equivalent to the 50th percentile or median
- `Q3`: third quartile, equivalent to the 75th percentile

The **IQR**, or interquartile range, measures the spread of the middle 50% of the data.

It is calculated as:

\[
IQR = Q3 - Q1
\]

The IQR is useful because it focuses on the central part of the distribution and is less affected by extremely high or extremely low values.

```python
selected_cols = [
    "Hours_Studied",
    "Previous_Scores",
    "Tutoring_Sessions",
    "Exam_Score"
]
```

```python
quartiles = df[selected_cols].quantile([0.25, 0.50, 0.75]).T

quartiles.columns = ["Q1", "Q2_median", "Q3"]

quartiles["IQR"] = quartiles["Q3"] - quartiles["Q1"]

quartiles
```

```python
percentile = (df["Exam_Score"] < student_score).mean() * 100

print(
    f"The student scored {student_score}. "
    f"This means the student scored higher than approximately {percentile:.2f}% "
    "of the students in the dataset."
)
```

## What Are Outliers?

Outliers are values that are very different from most of the other values in a variable.

They can happen for different reasons:

- a real but unusual case
- a data entry error
- a measurement error
- a special situation in the dataset

For example, if most students studied between 16 and 24 hours, a student who studied 44 hours may be considered unusual.

Outliers are important because they can affect statistical measures such as the mean and standard deviation. They can also influence machine learning models.

However, outliers should not be removed automatically. First, we need to investigate whether they are valid observations or data quality problems.

### Checking Outliers

Now we will check possible outliers using the IQR method.

The idea is to calculate:

IQR = Q3 - Q1

Then we define two limits:

Lower Limit = Q1 - 1.5 * IQR

Upper Limit = Q3 + 1.5 * IQR

Values below the lower limit or above the upper limit are considered possible outliers.

This method is the same logic used in a boxplot, but here we will only calculate the information without creating a graph.

```python
outlier_results = []

for col in selected_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    below = (df[col] < lower_limit).sum()
    above = (df[col] > upper_limit).sum()
    total = below + above

    outlier_results.append([
        col,
        Q1,
        Q3,
        IQR,
        lower_limit,
        upper_limit,
        below,
        above,
        total
    ])

outlier_summary = pd.DataFrame(
    outlier_results,
    columns=[
        "Variable",
        "Q1",
        "Q3",
        "IQR",
        "Lower Limit",
        "Upper Limit",
        "Below Limit",
        "Above Limit",
        "Total Outliers"
    ]
)

outlier_summary
```

After counting the outliers, we can inspect the actual values.

This helps us understand whether the outliers are extreme values, possible errors, or valid observations.

### Other Ways to Calculate Outliers

The IQR method is based on quartiles, but there are other ways to identify values that are very different from the rest of the data.

One common alternative is the Z-score method.

The Z-score measures how many standard deviations a value is from the mean.

Formula:

z = (x - mean) / standard deviation

Where:

- x is the value
- mean is the average of the variable
- standard deviation measures how spread out the values are

A common rule is:

- if z > 3, the value may be an unusually high outlier
- if z < -3, the value may be an unusually low outlier

This method works better when the variable has a distribution close to normal.

```python
zscore_results = []

for col in selected_cols:
    mean = df[col].mean()
    std = df[col].std()

    z_scores = (df[col] - mean) / std

    outliers = df[(z_scores < -3) | (z_scores > 3)]

    zscore_results.append([
        col,
        mean,
        std,
        outliers.shape[0]
    ])

zscore_summary = pd.DataFrame(
    zscore_results,
    columns=["Variable", "Mean", "Standard Deviation", "Total Outliers"]
)

zscore_summary
```

The IQR method uses quartiles and is less affected by extreme values. It is usually safer when the data is not normally distributed.

The Z-score method uses the mean and standard deviation. It is useful when the data is approximately normal, but it can be more sensitive to extreme values because both the mean and standard deviation are affected by outliers.

In practice, it is common to compare both methods instead of relying on only one.

### Business Rule Outliers

Sometimes, outliers should be detected using domain rules instead of statistical formulas.

For example, if `Exam_Score` should range from 0 to 100, then any value above 100 is invalid, even if it is not considered an outlier by a statistical method.

In this case, the value 101 should be checked because it may violate the expected rule for exam scores.

```python
df.loc[df["Exam_Score"] > 100, "Exam_Score"] = 100
```

## Correlation

Correlation measures the relationship between two numerical variables.

It helps us understand whether variables tend to increase or decrease together.


The correlation value ranges from -1 to 1:

- close to 1: strong positive relationship
- close to -1: strong negative relationship
- close to 0: weak or no linear relationship

Correlation does not mean causation. Even if two variables are correlated, it does not mean that one causes the other.

### Pearson and Spearman Correlation

Pearson and Spearman are two ways to measure correlation between variables.

Pearson measures a linear relationship.

- It works well when the variables change together in something close to a straight-line pattern.

- Example: as `Hours_Studied` increases, `Exam_Score` also increases in a mostly linear way.

Spearman measures a rank-based relationship.

- It checks whether higher values of one variable are generally associated with higher values of another variable, even if the relationship is not a straight line.

- Spearman is useful when the data has outliers or when the relationship is not linear.

In short:

Pearson = linear relationship  
Spearman = ranked or monotonic relationship

```python
correlation_matrix = df.corr(numeric_only=True, method="pearson")

correlation_matrix
```

```python
correlation_matrix = df.corr(numeric_only=True, method="spearman")

correlation_matrix
```

### Interpreting Correlation Strength

There is no perfect universal rule, but a common interpretation is:

- 0.00 to 0.19: very weak correlation
- 0.20 to 0.39: weak correlation
- 0.40 to 0.59: moderate correlation
- 0.60 to 0.79: strong correlation
- 0.80 to 1.00: very strong correlation

The same idea applies to negative values.

```python
exam_correlation = correlation_matrix["Exam_Score"].drop("Exam_Score").sort_values(ascending=False)

corr_comments = []

for variable, corr in exam_correlation.items():
    abs_corr = abs(corr)

    if abs_corr < 0.20:
        strength = "very weak"
    elif abs_corr < 0.40:
        strength = "weak"
    elif abs_corr < 0.60:
        strength = "moderate"
    elif abs_corr < 0.80:
        strength = "strong"
    else:
        strength = "very strong"

    if corr > 0:
        direction = "positive"
    elif corr < 0:
        direction = "negative"
    else:
        direction = "no"

    corr_comments.append([variable, corr, direction, strength])

corr_comments = pd.DataFrame(
    corr_comments,
    columns=["Variable", "Correlation with Exam Score", "Direction", "Strength"]
)

corr_comments
```

- The variable that stands out the most is `Attendance`. It has the strongest relationship with `Exam_Score`, with a correlation of 0.67. This suggests that attendance is much more associated with exam performance than the other variables.

- `Hours_Studied` also has a relevant relationship with `Exam_Score`, but it is weaker than attendance. This is interesting because study time matters, but it does not appear to explain exam performance as strongly as class attendance.

- `Previous_Scores` has a surprisingly low correlation with `Exam_Score`. We might expect previous academic performance to be strongly related to the final exam score, but in this dataset the relationship is very weak.

- `Tutoring_Sessions` also shows a very weak correlation. This may indicate that simply having more tutoring sessions is not strongly associated with higher exam scores, at least in a linear way.

- `Physical_Activity` and `Sleep_Hours` have almost no correlation with `Exam_Score`. In this dataset, they do not seem to have a meaningful relationship with exam performance.

## Hypothesis Testing

### Hypothesis Test 1: Average Attendance

Scenario:

The school claims that the average student attendance is higher than 80%.

We want to test whether the data provides statistical evidence to support this claim.

H0: The average Attendance is less than or equal to 80.

H1: The average Attendance is higher than 80.

Test:

One-sample t-test, right-tailed.

https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.ttest_1samp.html

```python
from scipy import stats
```

```python
reference_value = 80

t_stat, p_value_two_tailed = stats.ttest_1samp(
    df["Attendance"],
    reference_value,
    alternative="greater"
)

if t_stat > 0:
    p_value = p_value_two_tailed / 2
else:
    p_value = 1 - (p_value_two_tailed / 2)

print("T statistic:", t_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H0.")
    print("There is statistical evidence that the average Attendance is higher than 80%.")
else:
    print("Do not reject H0.")
    print("There is not enough evidence that the average Attendance is higher than 80%.")
```

### Hypothesis Test 2: Attendance and Exam Score

Scenario:

Do students with higher attendance have higher exam scores?

We want to test whether there is a statistically significant positive relationship between `Attendance` and `Exam_Score`.

H0: There is no positive linear relationship between Attendance and Exam_Score.

H1: There is a positive linear relationship between Attendance and Exam_Score.

Test:

Pearson correlation test, right-tailed.

https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.pearsonr.html

```python
corr, p_value_two_tailed = stats.pearsonr(
    df["Attendance"],
    df["Exam_Score"],
    alternative="greater"
)

if corr > 0:
    p_value = p_value_two_tailed / 2
else:
    p_value = 1 - (p_value_two_tailed / 2)

print("Pearson correlation:", corr)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H0.")
    print("There is statistical evidence of a positive relationship between Attendance and Exam_Score.")
else:
    print("Do not reject H0.")
    print("There is not enough evidence of a positive relationship between Attendance and Exam_Score.")
```

## Standardization and Normalization

Some numerical variables can have very different scales.

For example:

- `Attendance` goes from 60 to 100
- `Hours_Studied` goes from 1 to 44
- `Exam_Score` goes from 55 to 101

Many machine learning models are sensitive to scale. When variables have very different ranges, some variables may influence the model more simply because their numbers are larger.

Two common ways to rescale numerical variables are:

Standardization

Normalization

### Standardization

Standardization transforms a variable so that it has:

- mean equal to 0
- standard deviation equal to 1

Formula:

standardized value = (value - mean) / standard deviation

```python
from sklearn.preprocessing import StandardScaler

numeric_cols = [
    "Hours_Studied",
    "Attendance",
    "Sleep_Hours",
    "Previous_Scores",
    "Tutoring_Sessions",
    "Physical_Activity",
    "Exam_Score"
]

scaler = StandardScaler()

df_standardized = df.copy()

df_standardized[numeric_cols] = scaler.fit_transform(df[numeric_cols])
```

```python
df_standardized[numeric_cols].describe().T
```

### Normalization

Normalization transforms a variable to a fixed range.

The most common range is from 0 to 1.

Formula:

normalized value = (value - minimum) / (maximum - minimum)

```python
from sklearn.preprocessing import MinMaxScaler

normalizer = MinMaxScaler()

df_normalized = df.copy()

df_normalized[numeric_cols] = normalizer.fit_transform(df[numeric_cols])

df_normalized[numeric_cols].describe().T
```
