## Check Inconsistencies based on business rules on the following table

| Column    | Business rule                                                                              | Example of invalid value                                         |
| --------- | ------------------------------------------------------------------------------------------ | ---------------------------------------------------------------- |
| Email     | Must follow the pattern `first_name.last_name@example.com`                                 | `john@example.com`, `john.smith@gmail.com`, `jsmith@example.com` |
| Salary    | Must be within the accepted company salary range, for example between `52000` and `200000` | `5000`, `999999`, negative values                                |
| Age       | Must be within a realistic working-age range, for example between `18` and `70`            | `12`, `95`, negative values                                      |
| Join Date | Must be before today and after `2019`, the company foundation year                         | future dates, `2018-05-10`                                       |


```python
# Check and correct if needed
```

```python
today = pd.Timestamp.today().normalize()
```

```python
# 1. Email must be first_name.last_name@example.com

invalid_email = df[
    df["Email"] != (
        df["First_Name"].str.lower() + "." + df["Last_Name"].str.lower() + "@example.com"
    )
]

invalid_email
```

```python
# 2. Salary must be within the accepted range

invalid_salary = df[
    ~df["Salary"].between(52000, 200000)
]

invalid_salary
```

They need to get a raise, per minimum salary of the company. Maybe we could do a flag and report to the business

```python
minimum_salary = 52000

df.loc[df["Salary"] < minimum_salary, "New_Salary"] = minimum_salary

df.loc[df["Salary"] < minimum_salary, "Raise"] = (
    df["New_Salary"] - df["Salary"]
)

# Data to send to the HR
df.loc[df["Salary"] < minimum_salary, ["Employee_ID", "Raise"]]
```

```python
# 3. Age must be within the accepted range

invalid_age = df[
    ~df["Age"].between(18, 70)
]

invalid_age
```

```python
# 4. Join Date must be after 2019 and before today

invalid_join_date = df[
    ~df["Join_Date"].between(
        pd.Timestamp("2019-01-01"),
        today,
        inclusive="both"
    )
]

invalid_join_date
```

## Extract and create new column

```python
# Make a column for company time, based on todays date and join date
df["Company_Time"] = (
    today - pd.to_datetime(df["Join_Date"], errors="coerce")
).dt.days
```
